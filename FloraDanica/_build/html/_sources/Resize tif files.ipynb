This script resizes and stores tif files from a source folder to a destination folder. It first ensures the destination folder exists and then collects all TIFF files from the source. The script processes up to a specified number of these files, resizing each one to fit within a specified maximum size while maintaining its aspect ratio. The resized images are then saved in the destination folder in TIFF format.

```python
import os
from PIL import Image

def resize_and_store_tiff_files(source_folder, destination_folder, new_size=(100, 100), num_files=10):
    # Ensure the destination folder exists
    os.makedirs(destination_folder, exist_ok=True)

    # Get a list of all TIFF files in the source folder
    tiff_files = [f for f in os.listdir(source_folder) if f.lower().endswith('.tif') or f.lower().endswith('.tiff')]

    # Process only the first 'num_files' TIFF files
    for tiff_file in tiff_files[:num_files]:
        # Construct full file path
        source_path = os.path.join(source_folder, tiff_file)
        destination_path = os.path.join(destination_folder, tiff_file)

        # Open, resize, and save the image
        with Image.open(source_path) as img:
            # Resize the image while maintaining aspect ratio
            img.thumbnail(new_size, Image.LANCZOS)
            img.save(destination_path, "TIFF")


 
source_folder = r'./floradanica_0001-0250'
destination_folder = r'./floradanica_mini_sample_set'
resize_and_store_tiff_files(source_folder, destination_folder, new_size=(500, 500))
```
